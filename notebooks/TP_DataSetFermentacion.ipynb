## 1. Descripción general

La fermentación etanólica es una etapa fundamental en la producción industrial de bioetanol a partir de materias primas azucaradas, como el jugo y los derivados de la caña de azúcar. Su desempeño depende de múltiples factores operativos y biológicos, entre ellos la temperatura, la concentración inicial de azúcares, la viabilidad de las levaduras y los niveles de contaminación bacteriana. El seguimiento de estas variables permite evaluar el comportamiento del proceso e identificar condiciones asociadas con variaciones en la producción de etanol y el rendimiento fermentativo.

En este proyecto se aborda el análisis de datos de fermentaciones etanólicas industriales mediante un conjunto de datos sintéticos, diseñado para representar diferentes escenarios de operación. El dataset incluye registros de distintas plantas, sustratos y tratamientos de control bacteriano, así como variables fisicoquímicas, microbiológicas y productivas.

El trabajo comprende la preparación y limpieza de los datos, el análisis exploratorio de las variables y sus relaciones, y el desarrollo de visualizaciones interactivas que facilitan la interpretación de los resultados.

## 2. Objetivo del análisis

Explorar el comportamiento de variables relevantes del proceso de fermentación etanólica industrial e identificar patrones y asociaciones que puedan contribuir a comprender los factores relacionados con la viabilidad de las levaduras, la contaminación bacteriana y el rendimiento fermentativo.

En particular, se busca:

- Analizar la relación entre la contaminación bacteriana y la viabilidad de las levaduras.
- Comparar los niveles de contaminación y viabilidad asociados con diferentes tratamientos de control bacteriano.
- Explorar la relación entre la concentración inicial de azúcares (°Brix) y la producción final de etanol.
- Evaluar la asociación entre la temperatura de fermentación y el rendimiento fermentativo.
- Caracterizar la distribución de los rendimientos y su evolución temporal.
- Facilitar la comparación de resultados según la planta industrial y el tipo de sustrato utilizado.

###a.Normalización de nombres de las columnas

````python
# 1. Limpiamos espacios en blanco al inicio y al final que puedan tener las columnas
dff.columns = dff.columns.str.strip()

# 2. Definimos un diccionario de mapeo con nombres normalizados, simples y con unidades claras
nuevos_nombres = {
    'ID_registro': 'id_registro',
    'fecha_fermentacion': 'fecha',
    'planta': 'planta',
    'nro_lote': 'lote',
    'Tipo_Sustrato': 'sustrato',
    'tratamIento': 'tratamiento',
    'brix_inical': 'brix_inicial',
    'pH Inicial': 'ph_inicial',
    'temperatura_C': 'temp_c',
    'duracion_h': 'duracion_h',
    'viabilidad_levaduras_pct': 'viabilidad_pct',
    'Contaminacion_log_UFC_mL': 'contaminacion_log_ufc_ml',
    'ART_residual_g_L': 'arrt_gl',
    'etanol_final_pct_vv': 'etanol_final_pct_vv',
    'Rendim_Fermentativo_pct': 'rendimiento_pct'
}

# 3. Aplicamos el renombramiento en el DataFrame
dff2= dff.rename(columns=nuevos_nombres)

# Mostramos las nuevas columnas
print("Nuevas columnas normalizadas:")
print(list(dff.columns))
````

###b.Normalización de nombres de las variables categóricas


````python
# 1. Limpiamos espacios adicionales al inicio, al final y espacios dobles internos
dff2['tratamiento'] = dff2['tratamiento'].str.strip().str.replace(r'\s+', ' ', regex=True)

# 2. Convertimos el texto a un formato estándar (por ejemplo, Tipo Título o Minúsculas)
# Para evitar problemas de acentuación inconsistente y mayúsculas, mapeamos explícitamente a las categorías correctas
mapeo_tratamientos = {
    'Antibiótico': 'antibiótico',
    'antibiótico': 'antibiótico',
    'Antibiotico': 'antibiótico',
    'Sin tratamiento': 'sin tratamiento',
    'SIN TRATAMIENTO': 'sin tratamiento',
    'sin Tratamiento': 'sin tratamiento',
    'Extracto vegetal experimental': 'extracto vegetal experimental',
    'Extracto Vegetal Experimental': 'extracto vegetal experimental',
    'Ext. vegetal experimental': 'extracto vegetal experimental',
    'Biocida comercial': 'biocida comercial',
    'Biocida  comercial': 'biocida comercial'
}

# Aplicamos el mapeo para corregir todas las variantes
dff2['tratamiento'] = dff2['tratamiento'].map(mapeo_tratamientos).fillna(dff2['tratamiento'])

# 3. Verificamos cómo quedaron las frecuencias ahora
print("Frecuencia de tratamientos normalizada:")
print(dff2['tratamiento'].value_counts(dropna=False))

````

````python
# 1. Limpiamos espacios adicionales al inicio y al final en la columna sustrato
dff2['sustrato'] = dff2['sustrato'].str.strip().str.replace(r'\s+', ' ', regex=True)

# 2. Definimos el mapeo para unificar las categorías de sustratos
mapeo_sustratos = {
    'Melaza': 'melaza',
    'MELAZA': 'melaza',
    'Melza': 'melaza',
    'melaza': 'melaza',
    'Jugo de caña': 'jugo de caña',
    'JUGO DE CAÑA': 'jugo de caña',
    'jugo de Cana': 'jugo de caña',
    'Mezcla jugo-melaza': 'mezcla jugo-melaza',
    'mezcla jugo / melaza': 'mezcla jugo-melaza',
    'Miel final': 'miel final',
    'Miel  final': 'miel final'
}

# Aplicamos el mapeo para estandarizar
dff2['sustrato'] = dff2['sustrato'].map(mapeo_sustratos).fillna(dff2['sustrato'])

# 3. Verificamos las frecuencias normalizadas de sustratos
print("Frecuencia de sustratos normalizada:")
print(dff2['sustrato'].value_counts(dropna=False))
````

###b.Tratar los datos anómalos identificados anteriormente

````python
# Definimos los límites usando los nombres de columnas ya normalizados de dff2
limites_normalizados = {
    "brix_inicial": (0, None),
    "ph_inicial": (0, 8),
    "arrt_gl": (0, 30),
    "viabilidad_pct": (0, 100),
    "rendimiento_pct": (0, 100),
    "contaminacion_log_ufc_ml": (0, None)}

# Reemplazar los valores fuera de rango por NaN en dff2, para evitar que esos datos anómalos participen en el cálculo de estadísticos (media y/o mediana)
for columna, (minimo, maximo) in limites_normalizados.items():

    # Asegurar que la columna sea numérica
    dff2[columna] = pd.to_numeric(
        dff2[columna], errors="coerce" )

    # Identificar valores fuera de rango
    anomalos = dff2[columna] < minimo

    if maximo is not None:
        anomalos = anomalos | (dff2[columna] > maximo)

    # Reemplazar por NaN
    dff2.loc[anomalos, columna] = np.nan

    # MOSTRAR CUÁNTOS NaN HAY DESPUÉS
print("\nVALORES FALTANTES DESPUÉS DE LA LIMPIEZA")

display(dff2[list(limites_normalizados.keys())].isna().sum())
````

````python
# Guardar una copia antes de imputar
dff2_antes = dff2.copy()

#Voy a calcular las medianas por grupos, en función de cada variable, para hacer una imputación/reemplazo mas correcto de datos anómalos

# 1. Crear categorías de temperatura para el rendimiento (usando 'temp_c' que es el nombre correcto)
dff2["grupo_temperatura"] = pd.cut(
    dff2["temp_c"],
    bins=[-np.inf, 30, 35, np.inf],
    labels=["Baja", "Media", "Alta"])

# 2. Definir criterios de agrupación para cada variable (usando 'sustrato' en lugar de 'tipo_sustrato')
criterios = {
    "brix_inicial": ["sustrato"],
    "ph_inicial": ["sustrato", "planta"],
    "arrt_gl": ["sustrato"],
    "viabilidad_pct": ["planta"],
    "rendimiento_pct": ["planta", "sustrato", "grupo_temperatura"],
    "contaminacion_log_ufc_ml": ["planta", "tratamiento"]
}

# 3. Imputar los valores faltantes usando medianas por grupo
for columna, grupos in criterios.items():

    # Calcular la mediana correspondiente a cada grupo
    medianas = dff2.groupby(grupos, observed=True)[columna].transform("median")

    # Completar solamente los valores faltantes
    dff2[columna] = dff2[columna].fillna(medianas)

# 4. Comparar cantidad de faltantes antes y después
columnas = list(criterios.keys())

resumen = pd.DataFrame({
    "Faltantes antes": dff2_antes[columnas].isna().sum(),
    "Faltantes despues": dff2[columnas].isna().sum()
})

resumen["Valores imputados"] = (resumen["Faltantes antes" ] - resumen["Faltantes despues"])

print("RESUMEN DE IMPUTACIÓN")
display(resumen)

# 5. Mostrar las filas donde se imputaron valores
for columna in columnas:

    mascara = (dff2_antes[columna].isna() & dff2[columna].notna())

    comparacion = pd.DataFrame({
        "Planta": dff2.loc[mascara, "planta"],
        "Sustrato": dff2.loc[mascara, "sustrato"],
        "Valor anterior": dff2_antes.loc[mascara, columna],
        "Valor imputado": dff2.loc[mascara, columna]
    })

    print(f"\nVariable: {columna}")
    print(f"Cantidad de valores imputados: {len(comparacion)}")

    if len(comparacion) > 0:
        display(comparacion.head())

# 6. Verificar cuántos valores faltantes permanecen
print("\nVALORES FALTANTES DESPUÉS DE LA IMPUTACIÓN")
display(dff2[columnas].isna().sum())
````

> ⚠️ **Correlación no es causalidad.** Que dos cosas se muevan juntas no significa que una cause la otra.

### 5.6 📊 Mapa de calor (heatmap) de correlación

**❓ ¿Qué variables numéricas se mueven juntas?**  → 🧠 varias numéricas → **mapa de calor**

En vez de calcular la correlación de a pares, la calculamos para **todas a la vez** y la pintamos.

````python
corr = data_set_limpio.corr(numeric_only=True)    # Se agrega .corr para calcular la matriz de correlación
corr.round(2)
````

````python

plt.figure(figsize=(12, 9))
sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="RdBu_r",
    vmin=-1,
    vmax=1,
    center=0)
plt.title("Matriz de correlación entre variables numéricas")
plt.tight_layout()
plt.show()
````

📖 **Cómo leerlo:** cada casilla representa el coeficiente de correlación entre dos variables. Los tonos rojos indican correlaciones positivas y los azules, negativas. Cuanto más intenso es el color, mayor es la fuerza de la asociación lineal. La diagonal presenta siempre un valor de 1, porque representa la correlación de cada variable consigo misma.

Entre las relaciones más destacadas se observan **correlaciones positivas entre contaminación bacteriana y azúcares residuales (r = 0,73) y entre °Brix inicial y etanol final (r = 0,66)**. También aparecen **correlaciones negativas importantes entre azúcares residuales y rendimiento fermentativo (r = −0,79), contaminación bacteriana y rendimiento (r = −0,66), y contaminación bacteriana y viabilidad de levaduras (r = −0,55).**

Estas asociaciones permiten identificar patrones relevantes en el conjunto de datos, aunque no implican necesariamente relaciones de causa y efecto.

````python
# Automatizamos la generación de un archivo pdf con todo el análisis realizado
import subprocess
import sys

# Instalamos reportlab si no está en el entorno para poder generar el PDF
try:
    import reportlab
except ImportError:
    subprocess.check_call([sys.executable, "-m", "pip", "install", "reportlab"])

print("✅ Biblioteca reportlab lista para generar el PDF.")
````

````python

from datetime import datetime
from reportlab.lib.pagesizes import letter
from reportlab.platypus import (
    SimpleDocTemplate, Paragraph, Spacer,
    Table, TableStyle, PageBreak
)
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib import colors
from google.colab import files

def generar_reporte_pdf(filename="Reporte_Analisis_Fermentacion.pdf"):

    doc = SimpleDocTemplate(
        filename,
        pagesize=letter,
        rightMargin=40,
        leftMargin=40,
        topMargin=40,
        bottomMargin=40
    )

    story = []
    styles = getSampleStyleSheet()

    # ESTILOS
    titulo = ParagraphStyle(
        "TituloPersonalizado",
        parent=styles["Heading1"],
        fontSize=18,
        leading=23,
        textColor=colors.HexColor("#1A365D"),
        spaceAfter=14
    )

    subtitulo = ParagraphStyle(
        "SubtituloPersonalizado",
        parent=styles["Heading2"],
        fontSize=13,
        leading=17,
        textColor=colors.HexColor("#2B6CB0"),
        spaceBefore=12,
        spaceAfter=7
    )

    cuerpo = ParagraphStyle(
        "CuerpoPersonalizado",
        parent=styles["Normal"],
        fontSize=9.5,
        leading=14,
        textColor=colors.HexColor("#2D3748"),
        spaceAfter=8
    )

    # 1. TITULO
    story.append(Paragraph(
        "Reporte Ejecutivo de Análisis de Datos", titulo
    ))

    story.append(Paragraph(
        "<b>Caso de estudio:</b> Fermentación Etanólica Industrial",
        cuerpo
    ))

    story.append(Paragraph(
        f"<b>Fecha de generación:</b> {datetime.now().strftime('%d/%m/%Y')}",
        cuerpo
    ))

    story.append(Spacer(1, 12))

    # 2. INTRODUCCION
    story.append(Paragraph(
        "1. Introducción y Objetivos", subtitulo
    ))

    intro = (
        "La fermentación etanólica es una etapa fundamental en la "
        "producción industrial de bioetanol a partir de caña de azúcar. "
        "Este informe resume el análisis exploratorio de un conjunto "
        "de 6.000 registros sintéticos de fermentaciones industriales, "
        "con el objetivo de identificar patrones y asociaciones entre "
        "variables operativas, microbiológicas y productivas."
    )

    story.append(Paragraph(intro, cuerpo))

    # 3. CARACTERISTICAS DEL DATASET
    story.append(Paragraph(
        "2. Características del Dataset de Trabajo", subtitulo
    ))

    data_info = (
        f"El conjunto de datos procesado contiene "
        f"<b>{len(data_set_limpio):,}</b> registros y "
        f"<b>{len(data_set_limpio.columns)}</b> columnas, "
        "después de las etapas de normalización, identificación "
        "de duplicados, validación de variables y tratamiento "
        "de valores faltantes y fuera de rango. "
        "Durante la exploración posterior se identificaron "
        "algunas anomalías adicionales, como temperaturas "
        "incompatibles con las condiciones habituales del proceso."
    )

    story.append(Paragraph(data_info, cuerpo))

    # 4. CONCLUSIONES DEL ANALISIS
    story.append(Paragraph(
        "3. Conclusiones Clave del Análisis Exploratorio (EDA)",
        subtitulo
    ))

    preguntas_respuestas = [

        (
            "¿Existe correlación entre contaminación y viabilidad?",
            "Se identificó una correlación negativa moderada "
            "(r = -0,55). Los registros con mayor contaminación "
            "bacteriana tienden a presentar menores porcentajes "
            "de viabilidad de las levaduras."
        ),

        (
            "¿Qué tratamiento presenta menores niveles de contaminación?",
            "El tratamiento con antibiótico presentó los menores "
            "valores promedio y medianos de contaminación bacteriana "
            "(promedio de 4,11 log10 UFC/mL), seguido por el biocida "
            "comercial y el extracto vegetal experimental. "
            "El grupo sin tratamiento presentó los niveles más elevados."
        ),

        (
            "¿Existe relación entre azúcares iniciales y etanol final?",
            "Se observó una correlación positiva (r = 0,66). "
            "Los mostos con mayor concentración inicial de azúcares "
            "(grados Brix) tienden a presentar mayores concentraciones "
            "finales de etanol (% v/v)."
        ),

        (
            "¿Qué relación existe entre temperatura y rendimiento?",
            "Se identificó una correlación negativa muy débil "
            "(r = -0,11). No se observó una asociación lineal "
            "importante entre ambas variables en el conjunto "
            "de datos analizado."
        ),

        (
            "¿Existe correlación entre viabilidad y rendimiento?",
            "Se observó una correlación positiva (r = 0,62). "
            "Los registros con mayor viabilidad de levaduras "
            "tienden a presentar mayores rendimientos fermentativos."
        )
    ]

    for pregunta, respuesta in preguntas_respuestas:
        story.append(Paragraph(
            f"<b>{pregunta}</b>", cuerpo
        ))
        story.append(Paragraph(respuesta, cuerpo))
        story.append(Spacer(1, 3))

    # 5. TABLA ESTADISTICA
    story.append(PageBreak())

    story.append(Paragraph(
        "4. Resumen Estadístico de Variables Principales",
        subtitulo
    ))

    columnas = [
        "brix_inicial",
        "temp_c",
        "viabilidad_pct",
        "contaminacion_log_ufc_ml",
        "rendimiento_pct"
    ]

    stats_df = data_set_limpio[columnas].describe().round(2)

    # Nombres mas breves para facilitar la lectura
    nombres_columnas = [
        "Estadístico",
        "Brix inicial",
        "Temp. (C)",
        "Viabilidad (%)",
        "Contaminación<br/>log10(UFC/mL)",
        "Rendimiento (%)"
    ]

    etiquetas_filas = {
        "count": "Recuento",
        "mean": "Media",
        "std": "Desv. estándar",
        "min": "Mínimo",
        "25%": "25%",
        "50%": "Mediana",
        "75%": "75%",
        "max": "Máximo"
    }

    estilo_encabezado = ParagraphStyle(
        "EncabezadoTabla",
        fontName="Helvetica-Bold",
        fontSize=8,
        leading=10,
        textColor=colors.white,
        alignment=1
    )

    tabla_datos = [
        [Paragraph(t, estilo_encabezado) for t in nombres_columnas]
    ]

    for indice, fila in stats_df.iterrows():
        etiqueta = etiquetas_filas.get(indice, indice)

        valores = [etiqueta]

        for valor in fila:
            if indice == "count":
                valores.append(str(int(valor)))
            else:
                valores.append(f"{valor:.2f}")

        tabla_datos.append(valores)

    tabla = Table(
        tabla_datos,
        colWidths=[72, 76, 70, 91, 125, 98],
        repeatRows=1
    )

    tabla.setStyle(TableStyle([
        ("BACKGROUND", (0, 0), (-1, 0),
         colors.HexColor("#1A365D")),
        ("ALIGN", (0, 0), (-1, -1), "CENTER"),
        ("VALIGN", (0, 0), (-1, -1), "MIDDLE"),
        ("FONTNAME", (0, 1), (-1, -1), "Helvetica"),
        ("FONTSIZE", (0, 1), (-1, -1), 8),
        ("TOPPADDING", (0, 0), (-1, -1), 8),
        ("BOTTOMPADDING", (0, 0), (-1, -1), 8),
        ("ROWBACKGROUNDS", (0, 1), (-1, -1),
         [colors.white, colors.HexColor("#F0F4F8")]),
        ("GRID", (0, 0), (-1, -1), 0.4,
         colors.HexColor("#D7DEE6")),
    ]))

    story.append(tabla)
    story.append(Spacer(1, 18))

    # 6. LIMITACIONES
    story.append(Paragraph(
        "5. Alcance y Limitaciones", subtitulo
    ))

    limitaciones = (
        "El análisis se realizó sobre un dataset sintético "
        "generado con fines didácticos. Las asociaciones "
        "identificadas no permiten establecer relaciones "
        "causales ni extrapolar directamente los resultados "
        "a procesos industriales reales. Además, la presencia "
        "de algunos valores anómalos remanentes debe considerarse "
        "al interpretar las estadísticas y correlaciones."
    )

    story.append(Paragraph(limitaciones, cuerpo))

    story.append(Spacer(1, 10))

    story.append(Paragraph(
        "<i>Reporte generado automáticamente mediante Python "
        "y ReportLab a partir del dataset procesado en "
        "Google Colab.</i>",
        cuerpo
    ))

    # GENERAR PDF
    doc.build(story)

    print(f"PDF '{filename}' generado correctamente.")

# GENERAR Y DESCARGAR EL ARCHIVO
generar_reporte_pdf()
files.download("Reporte_Analisis_Fermentacion.pdf")
````
